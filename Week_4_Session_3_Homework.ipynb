## Task 1: Compare up to three countries

Use the Gapminder dataset and compare selected countries using interactive Plotly line charts.


````python
import plotly.express as px

df = px.data.gapminder()

countries = ["Uganda", "Kenya", "Tanzania"]

country_data = df[df["country"].isin(countries)]

country_data.head()
````

### Life Expectancy Over Time


````python
fig = px.line(
    country_data,
    x="year",
    y="lifeExp",
    color="country",
    markers=True,
    title="Life Expectancy Over Time"
)

fig.show()
````

### GDP per Capita Over Time


````python
fig = px.line(
    country_data,
    x="year",
    y="gdpPercap",
    color="country",
    markers=True,
    title="GDP per Capita Over Time"
)

fig.show()
````

## Task 2: Streamlit Country Explorer

The Streamlit app should:

- allow the user to select up to three countries;
- use a sidebar for the selector;
- use `st.cache_data` when loading data;
- display life expectancy and GDP per capita charts side by side.


````python
%%writefile mini_app.py
import streamlit as st
import plotly.express as px

st.set_page_config(
    page_title="Country Explorer",
    layout="wide"
)

st.title("Country Explorer")

@st.cache_data
def load_data():
    return px.data.gapminder()

df = load_data()

countries = st.sidebar.multiselect(
    "Select up to 3 countries",
    options=sorted(df["country"].unique()),
    default=["Uganda", "Kenya"],
    max_selections=3
)

if not countries:
    st.info("Select at least one country from the sidebar.")
    st.stop()

country_data = df[df["country"].isin(countries)]

col1, col2 = st.columns(2)

with col1:
    fig1 = px.line(
        country_data,
        x="year",
        y="lifeExp",
        color="country",
        markers=True,
        title="Life Expectancy Over Time"
    )

    st.plotly_chart(
        fig1,
        use_container_width=True
    )

with col2:
    fig2 = px.line(
        country_data,
        x="year",
        y="gdpPercap",
        color="country",
        markers=True,
        title="GDP per Capita Over Time"
    )

    st.plotly_chart(
        fig2,
        use_container_width=True
    )
````

### Run the Streamlit App

After running the cell above, start the app from a terminal using:

```bash
streamlit run mini_app.py
```


## Task 3: Plotly/Streamlit vs Matplotlib/Seaborn

Plotly and Streamlit are useful when we want interactive visualizations and a dashboard that users can explore. Plotly charts allow users to hover, zoom, pan, and interact with the data, while Streamlit provides widgets such as select boxes and multiselect controls that can change what is displayed. Matplotlib and Seaborn are more suitable for static visualizations, especially when preparing charts for reports or documents. Therefore, Plotly and Streamlit are a good choice for interactive applications, while Matplotlib and Seaborn are useful when a fixed chart is enough.
